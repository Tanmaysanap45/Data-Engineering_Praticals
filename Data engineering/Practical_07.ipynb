To design and implement an ETL pipeline using Python for extracting data from CSV files, cleaning and transforming the data, validating the records, and loading the processed data into a database.

```python
import pandas as pd
import sqlite3

# ============================================================
# 1. CSV ETL
# ============================================================

print("\n--- 1. CSV ETL Pipeline ---")

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Amit", "Neha", "Rahul"],
    "Age": [22, 25, 29],
    "Salary": [30000, 40000, 35000]
})

# Data Cleaning
df["Age"] = df["Age"].fillna(df["Age"].median())

# Data Transformation
df["Salary"] = df["Salary"] * 1.10

# Store in SQLite database
con = sqlite3.connect("etl.db")

df.to_sql(
    "employees",
    con,
    if_exists="replace",
    index=False
)

con.close()

print(df)


# ============================================================
# 2. MULTIPLE CSV FILES
# ============================================================

print("\n--- 2. Combining Multiple CSV Files ---")

a = pd.DataFrame({
    "ID": [1, 2],
    "Name": ["Amit", "Neha"]
})

b = pd.DataFrame({
    "ID": [3, 4],
    "Name": ["Rahul", "Priya"]
})

# Combine datasets
combined = pd.concat(
    [a, b],
    ignore_index=True
)

print(combined)


# ============================================================
# 3. JSON ETL
# ============================================================

print("\n--- 3. JSON ETL Pipeline ---")

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Amit", "Neha", "Rahul"],
    "Age": [22, 25, 28]
})

# Transformation
df["Age_Group"] = df["Age"].apply(
    lambda x: "Young" if x < 25 else "Adult"
)

# Store JSON
df.to_json(
    "users.json",
    orient="records"
)

# Store in SQLite
con = sqlite3.connect("users.db")

df.to_sql(
    "users",
    con,
    if_exists="replace",
    index=False
)

con.close()

print(df)


# ============================================================
# 4. REMOVE INVALID RECORDS
# ============================================================

print("\n--- 4. Invalid Record Removal ---")

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Age": [22, -5, 150]
})

print("Original Records:")

print(df)

# Remove invalid age records
df = df[
    (df["Age"] >= 0) &
    (df["Age"] <= 100)
]

print("\nAfter Removing Invalid Records:")

print(df)


# ============================================================
# 5. DATA VALIDATION
# ============================================================

print("\n--- 5. Data Validation ---")

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Age": [22, 25, 150]
})

# Validate age
valid = df[
    (df["Age"] >= 18) &
    (df["Age"] <= 100)
]

# Store validated records
con = sqlite3.connect("validated.db")

valid.to_sql(
    "employees",
    con,
    if_exists="replace",
    index=False
)

con.close()

print("Validated Records:")

print(valid)


# ============================================================
# 6. INCREMENTAL LOADING
# ============================================================

print("\n--- 6. Incremental Data Loading ---")

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Amit", "Neha", "Rahul"]
})

con = sqlite3.connect("incremental.db")

# Check whether table already exists
old = pd.read_sql(
    "SELECT name FROM sqlite_master "
    "WHERE type='table'",
    con
)

if old.empty:

    # First load
    df.to_sql(
        "employees",
        con,
        index=False
    )

else:

    # Append new records
    new = df

    new.to_sql(
        "employees",
        con,
        if_exists="append",
        index=False
    )

print("New Records Loaded:")

print(df)

con.close()


# ============================================================
# FINAL MESSAGE
# ============================================================

print("\n========== ALL ETL OPERATIONS COMPLETED ==========")
```
