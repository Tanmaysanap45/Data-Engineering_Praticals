To design and configure a scheduled Apache Airflow DAG for orchestrating an ETL workflow through sequential tasks for pipeline initialization, data extraction, and successful pipeline completion.

```python
# ============================================================
# APACHE AIRFLOW ETL PIPELINE PRACTICAL
# ============================================================

# 1. Install Apache Airflow
!pip install apache-airflow -q


# ============================================================
# 2. Import Libraries
# ============================================================

import os
import subprocess


# ============================================================
# 3. Configure Airflow
# ============================================================

os.environ["AIRFLOW_HOME"] = "/content/airflow"
os.environ["AIRFLOW__CORE__DAGS_FOLDER"] = "/content/airflow/dags"

os.makedirs("/content/airflow/dags", exist_ok=True)

print("Airflow environment configured successfully.")


# ============================================================
# 4. Create Data Extraction Python Script
# ============================================================

with open("/content/airflow/data_extraction.py", "w") as f:
    f.write("""
print("Starting data extraction...")
print("Extracting data from source...")
print("Data extraction completed successfully.")
""")

print("Data extraction script created successfully.")


# ============================================================
# 5. Create Airflow DAG
# ============================================================

with open("/content/airflow/dags/data_pipeline_dag.py", "w") as f:
    f.write("""
from datetime import datetime, timedelta

from airflow.sdk import DAG
from airflow.providers.standard.operators.bash import BashOperator
from airflow.providers.standard.operators.empty import EmptyOperator


# Default arguments
default_args = {
    "owner": "data_engineering_lab",
    "depends_on_past": False,
    "start_date": datetime(2026, 1, 1),
    "retries": 1,
    "retry_delay": timedelta(minutes=5),
}


# Create DAG
with DAG(
    dag_id="university_etl_orchestration",
    default_args=default_args,
    description="ETL pipeline orchestration",
    schedule="@daily",
    catchup=False,
) as dag:

    # Start task
    start_pipeline = EmptyOperator(
        task_id="start_pipeline"
    )

    # Extraction task
    execute_extraction = BashOperator(
        task_id="run_extraction_script",
        bash_command="python /content/airflow/data_extraction.py"
    )

    # Completion task
    pipeline_complete = BashOperator(
        task_id="log_pipeline_success",
        bash_command='echo "ETL Execution completed successfully"'
    )

    # Task dependency
    start_pipeline >> execute_extraction >> pipeline_complete
""")

print("DAG created successfully.")
print("DAG Name: university_etl_orchestration")
print("Schedule: Daily")


# ============================================================
# 6. Initialize Airflow Database
# ============================================================

print("\nInitializing Airflow database...\n")

db_result = subprocess.run(
    ["airflow", "db", "migrate"],
    env=os.environ,
    capture_output=True,
    text=True
)

print(db_result.stdout)

if db_result.stderr:
    print(db_result.stderr)


# ============================================================
# 7. Check DAG
# ============================================================

print("\nChecking Airflow DAG...\n")

dag_result = subprocess.run(
    [
        "airflow",
        "dags",
        "list"
    ],
    env=os.environ,
    capture_output=True,
    text=True
)

print(dag_result.stdout)

if dag_result.stderr:
    print(dag_result.stderr)


# ============================================================
# 8. Test Complete DAG
# ============================================================

print("\n==========================================")
print("Running DAG tasks...")
print("==========================================\n")

result = subprocess.run(
    [
        "airflow",
        "dags",
        "test",
        "university_etl_orchestration",
        "2026-08-20"
    ],
    env=os.environ,
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print(result.stderr)


# ============================================================
# 9. Practical Completed
# ============================================================

print("\n==========================================")
print("ETL Pipeline Practical Completed!")
print("==========================================")
```
